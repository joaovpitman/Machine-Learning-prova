````python
# --- Download e extração do acervo (Capítulo 1) ---
import requests, zipfile, io, os
import pandas as pd

url = "https://zenodo.org/records/21211053/files/openfcs_v1.0.0.zip?download=1"
resp = requests.get(url)
resp.raise_for_status()

with zipfile.ZipFile(io.BytesIO(resp.content)) as z:
    z.extractall("openfcs")

endereco = "openfcs/openfcs-1.0.0/data/derived/"
print("Download e extracao concluidos.")
````

### Questão 3 — SQL com DuckDB (2,5 pontos)

**a) (1,5 pt)** Escreva uma consulta SQL (via `duckdb.sql(...)`) que retorne, **para o ano de 2023**, o valor total exportado por domínio (`fcs_domain`), ordenado do maior para o menor.

**b) (1,0 pt)** Compare o resultado dessa consulta com o resultado equivalente obtido via `pandas` (`groupby`). Os dois batem? Mostre a comparação no código (não basta afirmar em texto — imprima os dois resultados lado a lado ou calcule a diferença).

````python
# Questao 3a e 3b — ESCREVA SEU CODIGO AQUI

````

````python
# Questão 3a: Consulta SQL para o valor total exportado por domínio em 2023
sql_query = """
SELECT
    fcs_domain,
    SUM(value_usd_millions) AS total_exported_value
FROM
    edges7
WHERE
    year = 2023
GROUP BY
    fcs_domain
ORDER BY
    total_exported_value DESC
"""
duckdb_result = duckdb.sql(sql_query).df()
print("Resultado SQL (2023 - Total Exportado por Domínio):")
display(duckdb_result)
````

````python
# Questão 3b: Comparação com resultado obtido via pandas
pandas_result = edges7[edges7['year'] == 2023].groupby('fcs_domain')['value_usd_millions'].sum().reset_index()
pandas_result = pandas_result.rename(columns={'value_usd_millions': 'total_exported_value'})
pandas_result = pandas_result.sort_values(by='total_exported_value', ascending=False).reset_index(drop=True)

print("\nResultado Pandas (2023 - Total Exportado por Domínio):")
display(pandas_result)

# Comparação dos resultados
comparison = duckdb_result.equals(pandas_result)
print(f"\nOs resultados do DuckDB e Pandas são idênticos? {comparison}")
````

````python
# Questão 4b: Consulta SQL com LEFT JOIN para os 10 maiores fluxos de exportação de 2024
sql_query_4b = """
SELECT
    e.economy,
    e.partner,
    e.fcs_domain,
    e.value_usd_millions,
    c.status AS mapping_status -- Corrigido de c.mapping_status para c.status
FROM
    edges7 AS e
LEFT JOIN
    crosswalk AS c ON e.cer_code = c.cer
WHERE
    e.year = 2024
ORDER BY
    e.value_usd_millions DESC
LIMIT 10
"""
duckdb_result_4b = duckdb.sql(sql_query_4b).df()
print("Top 10 fluxos de exportação em 2024 com status de mapeamento:")
display(duckdb_result_4b)
````
